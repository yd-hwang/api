## 정답

아래 셀을 실행하여 정답을 확인하세요.

````python
import requests

url = "https://api-k091.onrender.com/api/submit"

# ==========================================
# 정답 1: 기본 요청 보내기
# ==========================================
print("=" * 50)
print("정답 1: 기본 요청")
print("=" * 50)

response = requests.post(
    url,
    headers={
        "Content-Type": "application/json",
        "X-API-KEY": "suwon_univ_2026_api_key",
        "X-STUDENT-ID": "20201234"
    },
    json={"name": "홍길동", "course": "웹크롤자료분석"}
)
print(response.json())

# ==========================================
# 정답 2: 응답 상태코드 확인하기
# ==========================================
print("\n" + "=" * 50)
print("정답 2: 상태코드 확인")
print("=" * 50)

response = requests.post(
    url,
    headers={
        "Content-Type": "application/json",
        "X-API-KEY": "suwon_univ_2026_api_key",
        "X-STUDENT-ID": "20201234"
    },
    json={"name": "홍길동", "course": "웹크롤자료분석"}
)
print(f"상태코드: {response.status_code}")
print(f"응답: {response.json()}")

# ==========================================
# 정답 3: 헤더를 변수로 분리하기
# ==========================================
print("\n" + "=" * 50)
print("정답 3: 변수 분리")
print("=" * 50)

url = "https://api-k091.onrender.com/api/submit"
headers = {
    "Content-Type": "application/json",
    "X-API-KEY": "suwon_univ_2026_api_key",
    "X-STUDENT-ID": "20201234"
}
data = {"name": "홍길동", "course": "웹크롤자료분석"}

response = requests.post(url, headers=headers, json=data)
print(f"상태코드: {response.status_code}")
print(f"응답: {response.json()}")

# ==========================================
# 정답 4: 에러 상황 테스트
# ==========================================
print("\n" + "=" * 50)
print("정답 4: X-STUDENT-ID 누락")
print("=" * 50)

response = requests.post(
    url,
    headers={
        "Content-Type": "application/json",
        "X-API-KEY": "suwon_univ_2026_api_key"
    },
    json={"name": "홍길동", "course": "웹크롤자료분석"}
)
print(f"상태코드: {response.status_code}")
print(f"에러: {response.json()['detail']}")

# ==========================================
# 정답 5: GET 요청 에러 확인
# ==========================================
print("\n" + "=" * 50)
print("정답 5: GET 요청 (405 에러)")
print("=" * 50)

response = requests.get(url)
print(f"상태코드: {response.status_code}")
print(f"에러: {response.json()['detail']}")
````
